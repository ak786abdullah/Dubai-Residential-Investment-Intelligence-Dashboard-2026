```python
# --------------------------------------------------------------------------------
#                                          import libraries                       
# --------------------------------------------------------------------------------

import numpy as np 
import pandas as pd 
print("libraries imported successfully")
```

```python
# -----------------------------------------------------------------------------------
#                                             load data                              
# -----------------------------------------------------------------------------------

df=pd.read_csv("transactions_2024_to_2026.csv") 
print("data loaded successfully")
```

```python
# -------------------------------------------------------------------------------------
#                                             first look                               
# -------------------------------------------------------------------------------------

# show 1st five rows
display(df.tail(50))

# show total number of entries and columns,each column's data type ,non_null value's counts,memory_usage
display(df.info()) 
```

```python
# we have to analyze "Sell" ,"Sell-pre registration" and  "Residential" data 
# Extract the data 

df=df.query ("procedure_name == ['Sell','Sell - Pre registration'] and property_usage == 'Residential'")
```

```python
# Change instance_date data_type to datetime 
df["instance_date"]=pd.to_datetime(df["instance_date"],errors="coerce") 
print("data type changed successfuly")
```

```python
# Check duplicates 
display(df.duplicated().sum())
```

```python
# Check nulls 
null_values=df.isnull().sum()
total_values=df.shape[0]
print(f"{(null_values / total_values) * 100} %")
```

```python
# --------------------------------------------------------------------------------------------------
#                                                  handle null values 
# ----------------------------------------------------------------------------------------------------

df["property_sub_type"]=df["property_sub_type"].fillna("unknown") 

# Impute missing values using forward fill (ffill)
# Reason: The data follows a grouped structure where missing entries 
# imply a continuation of the preceding valid record (ditto marks).
df["building_name"]=df['building_name'].ffill(axis=0)
df['project_name']=df['project_name'].ffill(axis=0)
df['nearest_landmark']=df['nearest_landmark'].ffill(axis=0)
df['nearest_mall']=df['nearest_mall'].ffill(axis=0)
df['nearest_metro']=df['nearest_metro'].ffill(axis=0) 

# drop rooms NAN (~3%)
df=df.dropna(subset=["rooms"])
```

```python
df['rooms'].value_counts(dropna=False)
```

```python
# ensure the column is string type first 
df['rooms']=df['rooms'].astype(str)

# remove "B/R" 
df['rooms']=df['rooms'].str.replace("B/R",'',regex=True).str.strip()


# Single --> '1' 
df['rooms']=df['rooms'].str.replace('Single Room' ,'1',regex=True).str.strip()


# strip whitespace one last time 
df['rooms']=df['rooms'].str.strip().str.capitalize()

print(df['rooms'].value_counts())
```

```python
print(df.isnull().sum())
```

```python
# ------------------------------------------------------------------------------------------------------
#                                        Outliers_Remove 
# ------------------------------------------------------------------------------------------------------

# Define a fuction for outlier 
def remove_outliers_iqr(df, column):
    # Calculate Q1 (25th percentile) and Q3 (75th percentile)
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    
    # Calculate the Interquartile Range
    IQR = Q3 - Q1
    
    # Define bounds for outliers
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # Filter the DataFrame
    df_filtered = df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]
    
    return df_filtered

# Remove outlier 
df=remove_outliers_iqr(df,'procedure_area')
df=remove_outliers_iqr(df,'actual_worth')
df=remove_outliers_iqr(df,'meter_sale_price')

print(f'shape:{df.shape}')
```

```python
# procedure_area ,actual_worth and meter_sale_price must be positive 
# check for negitive values 
negitive_values=df.query("meter_sale_price <= 0 or actual_worth <= 0 or procedure_area <= 0") 
print(negitive_values.value_counts())
```

```python
# Check area unique_name
df['area_name'].unique()
```

```python
# ----------------------------------------------------------------------
#                                Standardize Area Names
# ----------------------------------------------------------------------

# Remove  the whitespace 

df['area_name']=df['area_name'].str.strip()

#  Define the Cleaning Map (Dictionary)
# This maps the "Wrong/Inconsistent Name" to the "Correct Standard Name"
cleaning_map = {
    # Fix "Barshaa" (extra 'a')
    'Al Barshaa South First': 'Al Barsha South First',
    'Al Barshaa South Second': 'Al Barsha South Second',
    'Al Barshaa South Third' : 'Al Barsha South Third',
    
    # Fix "Thanayah" (extra 'a')
    'Al Thanayah Fourth': 'Al Thanyah Fourth',
    
    # Fix "Kheeran" (This is the only merge: it is the same as Khairan)
    'Al Kheeran': 'Al Khairan First',
    
    # Standardize "Nadd" to "Nad"
    'Nadd Hessa': 'Nad Hessa'
}

#  Apply the fix
# We use .replace() which looks for exact matches of the keys and replaces them
df['area_name'] = df['area_name'].replace(cleaning_map)
```

```python
# -----------------------------------------------------------------------
#                           Standardize reg_type 
# -----------------------------------------------------------------------

df["reg_type"]=df["reg_type"] = df["reg_type"].map({
    "Off-Plan Properties": "Off-Plan",
    "Existing Properties": "Existing"
}).fillna(df["reg_type"]) # fillna keeps original values if not found in map
```

```python
# ------------------------------------------------------------------------------------
#                                       Feature_Engeering                            
# ------------------------------------------------------------------------------------

# convert area meter to feet
df["area_sqft"] = (df["procedure_area"]) * 10.764

# price per_sequare_feet
df["Price_Per_SqFt"] = df["actual_worth"] / df["area_sqft"]
```
