```python
# ---------------------------------------------------------
#                           import libraries
# ---------------------------------------------------------

import pandas as pd 
import numpy as np  
print("labriries imported successfuly")
```

```python
# ---------------------------------------------------------------
#                         load data 
# ----------------------------------------------------------------

df=pd.read_csv("rent_2025.csv") 
print("data loaded successfuly")
```

```python
# -----------------------------------------------------------
#                         first look 
# -----------------------------------------------------------

# show random 20 rows
display(df.sample(20))

```

```python
# standardize columns name 
df=df.rename(columns={"ejari_property_sub_type_en":"rooms","area_name_en":"area_name","annual_amount":"annual_rent","master_project_en":"master_project"})
```

```python
# check duplicate 
print(f"Number of duplicate row : {df.duplicated().sum()}")

# Remove duplicates 
df=df.drop_duplicates() 
print("duplicate rows are removed successfully")
```

```python
# check null values 
null_values=df.isnull().sum() 
total_values=df.shape[0]
print(f"number of null values : \n {(null_values / total_values) * 100 } % ") 

# remove null values ( ~ 0.5%) 
df=df.dropna() 
print('null values are removed successfully')
```

```python
# check rooms unique values
print(df["rooms"].unique())
```

```python
# -----------------------------------------------------------------------------------------------------------
#        
# ------------------------------------------------------------------------------------------------------------

# convert rooms data type into str 
df['rooms']=df['rooms'].astype(str)

# Extract digits and fill missing matches with the original value 
df['rooms']= df['rooms'].str.extract('(\\d+)',expand=False).fillna(df['rooms'])


df['rooms']= df['rooms'].str.strip().str.capitalize()
# note: i coverted rooms to  float becouse NaN are float they cannot convert to int  
# we need only residential data 

display(df.sample(20)) 
```

```python
# check null values 
df.isnull().sum()
```

```python
# ------------------------------------------------------------------------------------------------------
#                                        Outliers_Remove 
# ------------------------------------------------------------------------------------------------------

# Define a fuction for outlier 
def remove_outliers_iqr(df, column):
    # Calculate Q1 (25th percentile) and Q3 (75th percentile)
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    
    # Calculate the Interquartile Range
    IQR = Q3 - Q1
    
    # Define bounds for outliers
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # Filter the DataFrame
    df_filtered = df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]
    
    return df_filtered

# Remove outlier 
df=remove_outliers_iqr(df,'annual_rent')
```

```python
# Cleaning Dictionary
area_corrections = {
    # Fix Spelling/Transliteration
    'Al Barshaa South First': 'Al Barsha South First',
    'Al Barshaa South Second': 'Al Barsha South Second',
    'Al Barshaa South Third': 'Al Barsha South Third',
    'Al Thanayah Fourth': 'Al Thanyah Fourth',
    'Al Qoaz': 'Al Goze',
    'Nadd Hessa':'Nad Hessa',

    # Fix Caps (Case Sensitivity)
    'AL QUSAIS': 'Al Qusais',
    'AL TWAR FIFTH': 'Al Twar Fifth',
    'AL MIZHAR FOURTH': 'Al Mizhar Fourth',
    'OUD AL MUTEENA': 'Oud Al Muteena'
}

# replace area_name 
df['area_name'] = df['area_name'].replace(area_corrections)
```

```python
# --------------------------------------------------------------------------------
#                                Export cleaned_rent 
# -------------------------------------------------------------------------------

df.to_csv("cleaned_rent.csv",index=False)
```
