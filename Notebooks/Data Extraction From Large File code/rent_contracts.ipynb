```python
import pandas as pd 
import numpy as np
```

```python
file_path = 'Rent_Contracts.csv'
chunk_size = 2000000  # Adjust based on your RAM
filtered_list = []

# Iterate through the file in chunks
for chunk in pd.read_csv(file_path, chunksize=chunk_size, usecols=['contract_id','contract_start_date','annual_amount',
                                                                   'area_name_en','ejari_property_sub_type_en']):
    # Ensure the date column is in datetime format
    chunk['contract_start_date'] = pd.to_datetime(chunk['contract_start_date'],errors="coerce",format="%d-%m-%Y")
    
    # Filter for years 2022 to 2026
    mask = (chunk['contract_start_date'].dt.year >= 2025) & (chunk['contract_start_date'].dt.year <= 2026)
    filtered_chunk = chunk[mask]
    
    filtered_list.append(filtered_chunk)

# Combine all filtered chunks into one smaller DataFrame
final_df = pd.concat(filtered_list)
```

```python
final_df.to_csv("rent_2025.csv",index=False)
```
