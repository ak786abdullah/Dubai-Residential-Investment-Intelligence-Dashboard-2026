```python
import pandas as pd 
import numpy as np
```

```python
df=pd.read_csv("Transactions.csv") # load the data
print(f'raw data shape:{df.shape}')
```

```python
display(df.info()) # first look

# convert dtype of "instance_date" to datetime so thata we can extract data that is needed
df["instance_date"] =pd.to_datetime(df["instance_date"],errors="coerce",format="%d-%m-%Y")
```

```python
# we need data from  2022 to present and only sales data 
df=df.query("instance_date >= '01-01-2024' and trans_group_en =='Sales'")

# extract the columns that are needed
df=df[['transaction_id','trans_group_en','procedure_name_en','instance_date','property_type_en','property_sub_type_en'
       ,'property_usage_en','reg_type_en','area_name_en','building_name_en','project_name_en','nearest_landmark_en',
       'nearest_metro_en','nearest_mall_en','rooms_en','has_parking','procedure_area','actual_worth','meter_sale_price'
       ]]

# rename columns
df.columns=df.columns.str.replace('_en','')

# reset index 
df.reset_index(inplace=True,drop=True)

# shape of data after extraction 
print(f"data shape after data extraction:{df.shape}")
```

```python
df.to_csv("transactions_2024_to_2026.csv",index=False)
```
